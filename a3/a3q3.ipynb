```python
from gurobipy import Model, GRB

m = Model("a3q3")
x = m.addVars(4, lb=0.0, name="x")
c = [0.44, 0.20, 3.16, 0.72]
A = [[8.5, 1.1, 27, 2.5],   # Protein (g)
     [0,   2.4, 0,  2.3],   # Fibre (g)
     [15,  2,   5,  64],    # Vitamin A (%)
     [45,  0,   64, 0],     # Vitamin D (%)
     [30,  0,   0,  3]]     # Calcium (%)
b = [67, 30, 100, 100, 100]

m.setObjective(sum(c[j]*x[j] for j in range(4)), GRB.MINIMIZE)

constrs = []
names = ["Protein","Fibre","VitA","VitD","Calcium"]
for i in range(5):
    constrs.append(m.addConstr(sum(A[i][j]*x[j] for j in range(4)) >= b[i], name=names[i]))

m.optimize()

print("\nPrimal solution (servings):")
for j, food in enumerate(["milk","banana","salmon","broccoli"]):
    print(f"{food}: {x[j].X:.6f}")

print(f"\nMinimum cost: ${m.ObjVal:.6f}")
print("\nDuals (Pi) for [Protein, Fibre, VitA, VitD, Calcium]:")
for cn in constrs:
    print(f"{cn.ConstrName}: {cn.Pi:.8f}")

m.write("a3q3.lp")
```
