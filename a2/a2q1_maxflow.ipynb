### Q2A1 Maximum flow implementation

#### Original - 1b

```python
import gurobipy as gp
from gurobipy import GRB

nodes = ['s','a','b','c','d','e','t']
special = ['s','t']

arcs, cap = gp.multidict({
    ('s','a'): 15,
    ('s','b'): 6,
    ('s','c'): 4,
    ('a','b'): 3,
    ('a','d'): 10,
    ('b','d'): 8,
    ('b','e'): 14,
    ('c','b'): 5,
    ('c','e'): 6,
    ('d','t'): 7,
    ('e','d'): 12,
    ('e','t'): 18,
})

m = gp.Model('MaxFlow')
flow = m.addVars(arcs, name="flow")
m.setObjective(flow.sum('*', 't'),GRB.MAXIMIZE)
m.addConstrs((flow[a] <= cap[a] for a in arcs), name='cap')
m.addConstrs((flow.sum(n,'*')-flow.sum('*',n)==0 for n in 
              nodes if n not in special), name='conserve')
m.update()
m.write('a2q1_maxflow_b.lp')
m.optimize()
```

```python
m.printAttr('x')
```

#### Fractional - 1c

```python
nodes = ['s','a','b','c','d','e','t']
special = ['s','t']

arcs, cap = gp.multidict({
    ('s','a'): 15,
    ('s','b'): 6,
    ('s','c'): 4,
    ('a','b'): 3,
    ('a','d'): 7,
    ('b','d'): 8,
    ('b','e'): 9.2,
    ('c','b'): 0.2,
    ('c','e'): 3.8,
    ('d','t'): 7,
    ('e','d'): 12,
    ('e','t'): 18,
})

m = gp.Model('MaxFlow')
flow = m.addVars(arcs, name="flow")
m.setObjective(flow.sum('*', 't'),GRB.MAXIMIZE)
m.addConstrs((flow[a] <= cap[a] for a in arcs), name='cap')
m.addConstrs((flow.sum(n,'*')-flow.sum('*',n)==0 for n in 
              nodes if n not in special), name='conserve')
m.update()
m.write('a2q1_maxflow_c.lp')
m.optimize()
```

```python
m.printAttr('x')
```
